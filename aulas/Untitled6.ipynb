```python
from transformers import pipeline

classifier = pipeline("zero-shot-classification",
                      model="facebook/bart-large-mnli")

categorias = ["Política", "Economia", "Saúde", "Esportes", "Tecnologia"]

def classificar_com_zero_shot(titulo):
    resultado = classifier(titulo, candidate_labels=categorias)
    return resultado['labels'][0]  # categoria com maior score

# Teste
titulo = "Senado aprova nova reforma tributária"
print(classificar_com_zero_shot(titulo))
```
