# **&#9776; Tipos de Algoritmos de Classificação**

A escolha do algoritmo depende não apenas da natureza do problema e do tamanho do dataset, mas também da <b>necessidade de interpretabilidade, da precisão desejada e da capacidade computacional disponível.</b></h4>

# **&#9776; Árvore de Decisão**
---

<h4 align="justify">Quando falamos em "árvores", logo pensamos na floresta amazônica, em seus troncos robustos, raízes profundas e galhos que se estendem em múltiplas direções. No campo da Inteligência Artificial aplicada aos desafios socioambientais da Amazônia, essa imagem também encontra paralelo na técnica de Árvores de Decisão, um modelo de Machine Learning que, assim como as árvores da floresta, se ramifica em diferentes caminhos até alcançar um resultado final.</h4>
<h4 align="justify">As Árvores de Decisão funcionam como um fluxograma, composto por <b>nós de decisão</b> que representam regras baseadas nas variáveis disponíveis. Cada divisão organiza os dados em grupos mais homogêneos, até chegar aos <b>nós folha</b>, que expressam a previsão ou classificação. No contexto amazônico, podem ser aplicadas, por exemplo, para <b>classificar áreas em risco de desmatamento, identificar comunidades mais vulneráveis a eventos climáticos extremos ou prever se a qualidade da água de um rio atende a padrões de segurança.</b></h4>
<h4 align="justify">Embora sejam amplamente utilizadas em tarefas de <b>classificação</b>, as Árvores de Decisão também podem ser empregadas em <b>problemas de regressão</b>, como prever a variação de temperatura em determinada região ou estimar o volume de emissão de gases poluentes. Essa flexibilidade, aliada à clareza do modelo, torna as Árvores de Decisão uma ferramenta poderosa para apoiar decisões que integram ciência de dados e sustentabilidade, contribuindo para enfrentar de forma prática os desafios ambientais e sociais da Amazônia.</h4>

# **:: O Aprendizado na Árvore de Classificação**

<h4 align="justify">O aprendizado em uma Árvore de Classificação começa pelo princípio de que precisamos separar os dados em grupos que representem as classes de forma mais clara possível. O objetivo é reduzir a mistura entre as categorias e, assim, permitir que o modelo tome decisões mais precisas. Isso é feito avaliando os atributos disponíveis e testando como cada um deles pode dividir melhor os dados.</h4>
<h4 align="justify">No início do processo, o algoritmo considera todos os atributos do conjunto. Por exemplo, em um problema socioambiental, podemos ter variáveis como índice de desmatamento, proximidade de áreas de proteção ou nível de poluição da água. O modelo então avalia cada atributo para verificar qual deles possui maior poder de separação entre as classes de interesse.</h4>
<h4 align="justify">Para cada atributo, são testados diferentes pontos de corte. Esses pontos representam valores que podem ser usados para dividir os dados em dois ou mais grupos. Assim como no exemplo de empréstimos, em que a renda ou o valor solicitado podem ser utilizados como divisores, em problemas ambientais podemos testar limites de poluição ou limites de desmatamento.</h4>
<h4 align="justify">A escolha do ponto de corte ideal está diretamente ligada à impureza dos grupos formados. Impureza significa a presença de exemplos de classes diferentes dentro de um mesmo grupo. Um nó é considerado puro quando contém apenas exemplos de uma única classe, como “área preservada” sem registros de degradação. Quanto mais puro for o nó, melhor é a divisão.</h4>
<h4 align="justify">Para medir essa impureza, normalmente utiliza-se o índice de Gini, que varia de 0 a 1. Um valor de 0 representa pureza máxima (todos os exemplos são da mesma classe), enquanto valores mais próximos de 1 indicam nós muito misturados. Assim, o algoritmo procura reduzir o índice de Gini a cada nova divisão da árvore.</h4>
<h4 align="justify">Após calcular a impureza de todos os possíveis cortes em todos os atributos, o algoritmo escolhe aquele que melhor separa os dados, formando o nó raiz da árvore. Em seguida, repete o processo para os ramos subsequentes, criando novos nós de decisão até que as divisões não tragam ganhos relevantes ou até que seja atingido um critério de parada, como profundidade máxima.</h4>
<h4 align="justify">Esse processo de aprendizado resulta em uma árvore onde cada caminho representa uma sequência de decisões que leva a uma previsão final. No contexto socioambiental, isso significa poder prever, por exemplo, se uma região está em risco de desmatamento ou não, se a água de um rio está própria para consumo ou imprópria, ou se uma comunidade está em situação de alta ou baixa vulnerabilidade ambiental.</h4>

```python
# Carrega o arquivo dataset_socioambiental_60.csv em um DataFrame
df = pd.read_csv('/content/dataset_socioambiental_60.csv')
```

# **CRISP-DM :: PREPARAÇÃO DOS DADOS**

### Aplicando *Hold Out*

```python
# ==========================
# Fase de Preparação dos Dados - CRISP-DM
# Modelo: Árvore de Decisão
# ==========================

from sklearn.model_selection import train_test_split

# Separar features (X) e target (y)
X = df.drop('risco_socioambiental', axis=1)
y = df['risco_socioambiental']

# Aplicar Hold Out (80% treino, 20% teste)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("Dimensão do X_train:", X_train.shape)
print("Dimensão do X_test:", X_test.shape)
print("Dimensão do y_train:", y_train.shape)
print("Dimensão do y_test:", y_test.shape)
```

# **CRISP-DM :: MODELAGEM**

### **Criando um Modelo baseado em Árvore de Decisão**

```python
# ==========================
# Fase de Modelagem - CRISP-DM
# Modelo: Árvore de Decisão
# ==========================

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Criar e treinar o modelo
modelo = DecisionTreeClassifier(
    criterion="gini",      # ou "entropy"
    max_depth=5,           # limite de profundidade (ajustável)
    random_state=42
)

modelo.fit(X_train, y_train)

# Fazer previsões
y_pred = modelo.predict(X_test)

# Avaliar desempenho
print("Acurácia:", accuracy_score(y_test, y_pred))
```

<h4 align="justify"><b>observação:</b> Use <b>Gini</b> quando buscar rapidez e simplicidade, pois é menos custoso computacionalmente e gera resultados semelhantes. Prefira <b>Entropy</b> quando desejar maior rigor estatístico na escolha dos splits, especialmente em problemas acadêmicos ou que exigem interpretação informacional. Ambos normalmente produzem árvores muito parecidas em desempenho.</h4>

### **Visualização da Árvore de Decisão**

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

# Plotar a árvore de decisão
plt.figure(figsize=(20, 10))
plot_tree(
    modelo,
    feature_names=X_train.columns,   # nomes das variáveis preditoras
    class_names=[str(c) for c in modelo.classes_],  # nomes das classes
    filled=True,                     # cores para facilitar leitura
    rounded=True,                    # cantos arredondados
    fontsize=10
)
plt.show()
```

# **CRISP-DM :: AVALIAÇÃO**

### **Classificação**

```python
# ==========================
# Fase de Avaliação - CRISP-DM
# Modelo: Árvore de Decisão
# ==========================

from sklearn import metrics
import matplotlib.pyplot as plt

# Calcular a matriz de confusão
confusion_matrix = metrics.confusion_matrix(y_test, y_pred)

# Exibir a matriz de confusão usando ConfusionMatrixDisplay
cm_display = metrics.ConfusionMatrixDisplay(confusion_matrix = confusion_matrix, display_labels = [0, 1])

cm_display.plot()
plt.show()
```

**Valores obtidos:**

TN (Verdadeiros Negativos): 0

FP (Falsos Positivos): 2

FN (Falsos Negativos): 1

TP (Verdadeiros Positivos): 15

### **Resultados Explicados**

### **Matriz de Confusão**

<h4 align="justify">A Matriz de Confusão criada possui quatro quadrantes diferentes:</h4>


|                      | **Previsto Positivo** | **Previsto Negativo** |
|----------------------|-----------------------|-----------------------|
| **Real Positivo**    | VP (Verdadeiro Positivo) | FN (Falso Negativo)   |
| **Real Negativo**    | FP (Falso Positivo)      | VN (Verdadeiro Negativo) |

- **VP + VN** → acertos do modelo  
- **FP + FN** → erros do modelo  

<h4 align="justify"><b>Verdadeiro significa que os valores foram previstos com precisão. Falso significa que houve um erro ou uma previsão errada.</b></h4>

### **Métricas Criadas**

<h4 align="justify">As diferentes medidas incluem: acurácia, precisão, sensibilidade (recall), especificidade e o F1-score, explicados a seguir.</h4>

### **Acurácia**

<h4 align="justify">A acurácia é a métrica mais intuitiva para avaliar modelos de classificação. Ela mede a proporção de previsões corretas em relação ao total de casos analisados, considerando tanto acertos positivos quanto negativos.</h4>

$$
\text{Acurácia} = \frac{VP + VN}{VP + VN + FP + FN}
$$

```python
from sklearn.metrics import accuracy_score

# Calcular a acurácia
accuracy = accuracy_score(y_test, y_pred)

print(f"Acurácia: {accuracy:.4f}")
```

<h4 align="justify">Um valor alto de acurácia indica que o modelo acerta a maioria das previsões, mas pode ser enganoso em conjuntos de dados desbalanceados.</h4>

### **Precisão**

<h4 align="justify">Dos positivos previstos, qual porcentagem é verdadeiramente positiva?</h4>

$$
\text{Precisão} = \frac{VP}{VP + FP}
$$

```python
from sklearn.metrics import precision_score

# Calcular a precisão usando precision_score
precision = precision_score(y_test, y_pred)

print(f"Precisão: {precision:.4f}")
```

<h4 align="justify">Em cada 100 casos que o modelo marcou como positivo, 88 realmente eram positivos, e 12 eram falsos positivos.</h4>

### **Sensibilidade**

<h4 align="justify">De todos os casos positivos, qual a porcentagem prevista de positivos?</h4>
<h4 align="justify">A sensibilidade (às vezes <b>chamada de recall</b>) mede o quão bom o modelo é em prever resultados positivos.</h4>
<h4 align="justify">Isso significa que ele analisa verdadeiros positivos e falsos negativos (que são positivos que foram incorretamente previstos como negativos).</h4>

$$
\text{Sensibilidade/Recall} = \frac{VP}{VP + FN}
$$

```python
from sklearn.metrics import recall_score

# Calcular o recall usando recall_score
recall = recall_score(y_test, y_pred)

print(f"Sensibilidade/Recall: {recall:.4f}")
```

<h4 align="justify">Em cada 100 exemplos que eram de fato positivos, o modelo acertou cerca de 94 e errou apenas 6 (classificando-os como negativos por engano).</h4>

### **Especificidade**


<h4 align="justify">Quão bem o modelo é em prever resultados negativos?</h4>
<h4 align="justify">A especificidade é semelhante à sensibilidade, mas a analisa pela perspectiva de resultados negativos.</h4>

$$
\text{Especificidade} = \frac{VN}{VN + FP}
$$

```python
from sklearn.metrics import confusion_matrix

# carrega os valores da matriz de confusão
vn, fp, fn, vp = confusion_matrix(y_test, y_pred).ravel()

#calcula a especificidade
specificity = vn / (vn + fp)
print(vn)
print(vp)

print(f"Especificidade: {specificity:.4f}")
```

<h4 align="justify">Em outras palavras: o modelo “acerta” nos positivos, mas erra completamente os negativos.</h4>

### **F1-Score**

<h4 align="justify">F1-score é a "média harmônica" de precisão e sensibilidade.</h4>
<h4 align="justify">Ele considera casos de falso positivo e falso negativo e é bom para conjuntos de dados desbalanceados.</h4>

$$
F_{1} = 2 \cdot \frac{\text{Precisão} \cdot \text{Recall}}{\text{Precisão} + \text{Recall}}
$$


```python
from sklearn.metrics import f1_score

# Calcular o F1-score
f1 = f1_score(y_test, y_pred)

print(f"F1-Score: {f1:.4f}")
```

<h4 align="justify">Em cada 100 previsões positivas/negativas que o modelo faz, cerca de 91% refletem um bom equilíbrio entre acertos e erros.:</h4>

### **Guia rápido de métricas de classificação**

| Métrica        | Definição                                                                 | Quando usar                                        | Exemplo prático                          |
|----------------|---------------------------------------------------------------------------|---------------------------------------------------|------------------------------------------|
| **Acurácia**   | Proporção total de acertos (positivos + negativos corretos)               | Classes **balanceadas**                           | Classificar imagens de gatos e cachorros |
| **Precisão**   | Dos casos previstos como positivos, quantos realmente são positivos       | Quando **falsos positivos custam caro**           | Diagnóstico de câncer                    |
| **Recall**     | Dos casos realmente positivos, quantos foram corretamente identificados   | Quando **falsos negativos custam caro**           | Detecção de fraude bancária              |
| **Especificidade** | Dos casos realmente negativos, quantos foram corretamente identificados | Quando **falsos alarmes são perigosos**           | Controle de qualidade em fábrica         |
| **F1-Score**   | Equilíbrio entre Precisão e Recall (média harmônica)                      | Quando há **desequilíbrio de classes** e precisamos de **equilíbrio** | Filtro de e-mails spam                   |
